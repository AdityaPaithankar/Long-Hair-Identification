```python
from datasets import load_dataset

ds = load_dataset(
    "Zhincore/synthetic-human-portrait-attributes",
    split="train",
    streaming=True
)

print(next(iter(ds)))
```

```python
labels_ds = ds.remove_columns(["image"])

ages = set()
hair_lengths = set()
sexes = set()

for row in labels_ds:
    ages.add(row["age"])
    hair_lengths.add(row["hair_length"])
    sexes.add(row["sex"])

print("AGE:", sorted(ages))
print("HAIR:", sorted(hair_lengths))
print("SEX:", sorted(sexes))
```
